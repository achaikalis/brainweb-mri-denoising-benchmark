# 00 - Dataset Download

Downloads the full BrainWeb *Normal* Phantom Grid (**T1 / T2 / PD x noise {0, 1, 3, 5, 7, 9}% x RF {0, 20, 40}% @ 1 mm = 54 volumes**) as **MINC** MRI Volumes.

Outputs: `data/raw/minc/<contrast>_ICBM_normal_1mm_pn<noise>_rf<rf>.mnc.gz` plus `data/raw/minc/manifest.csv` (sizes + SHA-256).

```python
import pandas as pd
from brainweb_denoising.download import fetch_all_minc
```

```python
import logging
logging.basicConfig(level=logging.INFO, format="%(levelname)s %(message)s")
```

```python
MINC_DIR = "data/raw/minc"
MANIFEST = "data/raw/minc/manifest.csv"
```

```python
# Use force=True to re-download existing valid files in MINC_DIR
rows = fetch_all_minc(MINC_DIR, manifest_path=MANIFEST, force=False)
print(f"{len(rows)} Volumes Present")
```

```python
import gzip

manifest = pd.read_csv(MANIFEST)
assert len(manifest) == 54, f"expected 54 rows, got {len(manifest)}"
assert manifest.groupby("contrast").size().eq(18).all()
for p in manifest["path"]:
    with gzip.open(p, "rb") as fh:
        assert fh.read(4).startswith(b"CDF"), f"not a MINC (netCDF) file: {p}"
print("All", len(manifest), "volumes are valid gzipped MINC files.")
manifest.head()
```
